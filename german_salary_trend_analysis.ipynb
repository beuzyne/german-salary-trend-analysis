#Historical Gross Monthly Earnings Trend in Germany (1991-2024)

```python
import pandas as pd
import matplotlib.pyplot as plt

#1. READ THE UPLOADED GERMAN DESTATIS FILE
file_name = "german_monthly_earnings.csv"
df_germany = pd.read_csv("german_monthly_earnings.csv")
```

markdown
## Step 2 : Inspecting the Dataset Structure and Dimensions
In this section, we check the total rows and columns of the Destatis dataset and look for any initial statistics

```python
#Check how many rows and columns (Dimensions)
print("---DATASET SHAPE (Rows, Columns)---")
print(df_germany.shape)

#Peek at the top 5 records of the dataset
print("\n---FIRST 5 ROWS OF GERMAN EARNINGS---")
print(df_germany.head())
```

markdown
## Step 3 : Detecting Missing Values in the Dataset
We use the `.isnull().sum()` function to check if there are any empty cells (NaN) in our German salary table.

```python
# Count missing values for each column
print("---MISSING VALUES COUNT PER COLUMN---")
print(df_germany.isnull().sum())
```

markdown
## Step 4: Data Cleaning Validation
Since the initial check returnd - missing values, we finalize the dataset preparation by ensuring a safe cleaned copy of the DataFrame is ready for analysis.

```python
#Create a safe cleaned copy of the dataset
df_clean_germany = df_germany.dropna()

print("---FINAL CLEANED DATASET SHAPE---")
print(df_clean_germany.shape)
```

markdown
##Step 5: Data Visualization - Salary Trends in Germany (1991-2024)
In this final section, we create a custom line chart using Matplotlib to visualize and compare the historical gross monthly earnings trend between men and women in Germany.

```python
import matplotlib.pyplot as plt

#Set the plot size for better visibility
plt.figure(figsize =(10,6))

#Plot lines for Men and Women salaries over the Years
#Use purple for Women and Blue for Man

plt.plot(df_clean_germany["Year"], df_clean_germany["Men"], label = "Men", color="royalblue", marker= "o")
plt.plot(df_clean_germany["Year"],df_clean_germany["Women"], label = "Women",color= "purple", marker = "o")

#Customing the chart labels and title
plt.title("Gross Monthly Earnings Trend in Germany (1991-2024)", fontsize=14, fontweight="bold")
plt.xlabel("Year", fontsize=12)
plt.ylabel("Average Salary in Euro (€)", fontsize=12)
plt.grid(True, linestyle="--", alpha=0.6)
plt.legend ()

#Display the final chart
plt.show()

```
